### Question 5

How many movies are there that are both in the "Thriller" genre and contain the word “love” anywhere in the keywords?


**Options:**
 - 48
 - 38
 - 14
 - 1

```python
%%sql
select * from movies limit 1;
```

```python
%%sql
select * from genres;
```

```python
%%sql
select * from genremap limit 5;
```

```python
%%sql
select * from keywords limit 5;
```

```python
%%sql
select * from keywordmap limit 5;
```

```python
%%sql
drop view if exists love_thrillers;
create view love_thrillers as
select DISTINCT
    gm.movie_id,
    m.title,
    m.original_title,
    m.release_date,
    m.overview,
    m.tagline,
    g.genre_name
from movies m
join genremap gm
    on m.movie_id = gm.movie_id
join genres g
    on gm.genre_id = g.genre_id
join keywordmap km
    on m.movie_id = km.movie_id
join keywords k
    on km.keyword_id = k.keyword_id
where 
    lower(k.keyword_name) like '%love%'
AND
    g.genre_name = 'Thriller'
order by m.release_date asc;
```

```python
%%sql
select * from love_thrillers limit 5;
```

```python
%%sql
select count(*) as total_love_thrillers
from love_thrillers;
```

How many movies are there that were released between 1 August 2006 ('2006-08-01') and 1 October 2009 ('2009-10-01') that have a popularity score of more than 40 and a budget of less than 50 000 000?

 
**Options:**

 - 29
 - 23
 - 28
 - 35

```python
%%sql
select * from movies limit 1;
```

```python
%%sql
drop view if exists recent_popular_low_budget_movies;
create view recent_popular_low_budget_movies as
select
    title,
    DATE(release_date) as release_date,
    popularity,
    budget
from movies
where 
    DATE(release_date) BETWEEN '2006-08-01' AND '2009-10-01'
AND
    popularity > 40
AND
    budget < 50000000
order by release_date asc;
```
