```python
import numpy as np
import pandas as pd
import cufflinks as cf
import cufflinks.colors as cufflinks_colors
from cufflinks import plotlytools
from sklearn import preprocessing
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split

from sklearn.ensemble import RandomForestRegressor

import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

sns.set_theme(style='whitegrid', palette='muted',
        rc={'figure.figsize': (15,10)})

from plotly.offline import download_plotlyjs, init_notebook_mode, plot, iplot
init_notebook_mode(connected=True)

if not hasattr(cufflinks_colors, '_original_to_rgba'):
    cufflinks_colors._original_to_rgba = cufflinks_colors.to_rgba

_original_to_rgba = cufflinks_colors._original_to_rgba

def _to_rgba_compat(color, alpha):
    return _original_to_rgba(color, float(alpha))

cufflinks_colors.to_rgba = _to_rgba_compat
plotlytools.to_rgba = _to_rgba_compat
```
